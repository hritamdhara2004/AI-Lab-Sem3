# IMPLEMENT MINIMAX ALGORITHM

```python
import math
import matplotlib.pyplot as plt
def minimax(depth, node_index, is_maximizing, values, max_depth):
    if depth == max_depth:
        return values[node_index]
    if is_maximizing:
        return max(
            minimax(depth + 1, node_index * 2, False, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, False, values, max_depth)
        )
    else:
        return min(
            minimax(depth + 1, node_index * 2, True, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, True, values, max_depth)
        )
leaf_values = [3, 5, 2, 9, 15, 5, 23, 23]
tree_depth = int(math.log2(len(leaf_values)))
best_score = minimax(0, 0, True, leaf_values, tree_depth)
print("Best possible score:", best_score)
positions = {
    0: (4, 3),
    1: (2, 2), 2: (6, 2),
    3: (1, 1), 4: (3, 1), 5: (5, 1), 6: (7, 1),
    7: (0.5, 0), 8: (1.5, 0),
    9: (2.5, 0), 10: (3.5, 0),
    11: (4.5, 0), 12: (5.5, 0),
    13: (6.5, 0), 14: (7.5, 0)
}
labels = {
    0: "MAX",
    1: "MIN", 2: "MIN",
    3: "MAX", 4: "MAX", 5: "MAX", 6: "MAX",
    7: "3", 8: "5",
    9: "2", 10: "9",
    11: "15", 12: "5",
    13: "23", 14: "23"
}

plt.figure(figsize=(12, 7))
for parent in range(7):
    left = 2 * parent + 1
    right = 2 * parent + 2

    plt.plot(
        [positions[parent][0], positions[left][0]],
        [positions[parent][1], positions[left][1]]
    )

    plt.plot(
        [positions[parent][0], positions[right][0]],
        [positions[parent][1], positions[right][1]]
    )
for node, (x, y) in positions.items():
    plt.scatter(x, y, s=1200)
    plt.text(
        x, y, labels[node],
        ha="center",
        va="center",
        fontsize=12
    )
plt.title("Minimax Game Tree")
plt.axis("off")
plt.show()
```

***BFS Traversal*** 

```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt

# Define the Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

# BFS Function
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    traversal = []
    visited.add(start)

    while queue:
        node = queue.popleft()
        traversal.append(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return traversal

# Run BFS
result = bfs(graph, 'A')

print("BFS Traversal:")
print(" -> ".join(result))

#Create a directed Geraph
G = nx.DiGraph()

# Add edges from dictionary
for node, neighbors in graph.items():
    for neighbor in neighbors:
        G.add_edge(node, neighbor)

#Define positions for better tree layout
pos= {
    'A':(0, 3),
    'B':(-1, 2),
    'C':(1, 2),
    'D':(-1.5, 1),
    'E':(-0.5 ,1),
    'F':(1,1),
    'G':(-0.5,0)
}
#DRAW THE GRAPH
plt.figure(figsize=(4,3))

nx.draw(
    G, pos,
    with_labels=True,
    node_color='skyblue',
    node_size=1250,
    font_size=10,
    font_weight='bold',
    edge_color='black',
    arrows=True,
    width=2
)
plt.title('BFS Graphical Representation',fontsize=16)
plt.axis('off')
plt.show()
```

```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt


# Define the Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}


# DFS Function (Recursive)
def dfs(graph, node, visited=None):
    if visited is None:
        visited = set()

    traversal = []

    visited.add(node)
    traversal.append(node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            traversal.extend(dfs(graph, neighbor, visited))

    return traversal


# Run DFS
result = dfs(graph, 'A')

print("DFS Traversal:")
print(" -> ".join(result))


# Create a directed Graph
G = nx.DiGraph()


# Add edges from dictionary
for node, neighbors in graph.items():
    for neighbor in neighbors:
        G.add_edge(node, neighbor)


# Define positions for better tree layout
pos = {
    'A': (0, 3),
    'B': (-1, 2),
    'C': (1, 2),
    'D': (-1.5, 1),
    'E': (-0.5, 1),
    'F': (1, 1),
    'G': (-0.5, 0)
}


# Draw the graph
plt.figure(figsize=(4, 3))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color='lightgreen',
    node_size=1250,
    font_size=10,
    font_weight='bold',
    edge_color='black',
    arrows=True,
    width=2
)


# Title
plt.title('DFS Graphical Representation', fontsize=16)
plt.axis('off')
plt.show()
```

# Rule-Based Expart system for student performance evaluation

```python
print("STUDENT PERFORMANCE EVALUATION")
print("-" * 40)
name = input("Enter student name: ")
attendance = float(input("Enter attendance percentage: "))
marks = float(input("Enter marks percentage: "))
assignment = float(input("Enter assignment performance percentage: "))
internal = float(input("Enter internal assessment percentage: "))

if attendance >= 90 and marks >= 85 and assignment >= 85 and internal >= 85:
    performance = "EXCELLENT"
elif attendance >= 75 and marks >= 70 and assignment >= 70 and internal >= 70:
    performance = "GOOD"
elif attendance >= 60 and marks >= 50 and assignment >= 50 and internal >= 50:
    performance = "AVERAGE"
else:
    performance = "POOR"
    
print("\n" + "-" * 40)
print("Student Name:", name)
print("Performance:", performance)
print("-" * 40)
```

# Rule-based expert system for medical diagnosis

```python
print("MEDICAL DIAGNOSIS EXPERT SYSTEM")
print("-" * 40)
fever = input("Do you have fever? (yes/no): ").lower()
cough = input("Do you have cough? (yes/no): ").lower()
headache = input("Do you have headache? (yes/no): ").lower()
sore_throat = input("Do you have sore throat? (yes/no): ").lower()

if fever == "yes" and cough == "yes" and sore_throat == "yes":
    disease = "Possible Flu"
elif fever == "yes" and headache == "yes" and cough == "no":
    disease = "Possible Viral Fever"
elif cough == "yes" and sore_throat == "yes" and fever == "no":
    disease = "Possible Common Cold"
elif fever == "yes" and cough == "yes" and headache == "yes" and sore_throat == "yes":
    disease = "Possible Flu"
else:
    disease = "Unknown - Further Medical Evaluation Required"

print("\n" + "-" * 40)
print("Possible Diagnosis:", disease)
print("-" * 40)
```

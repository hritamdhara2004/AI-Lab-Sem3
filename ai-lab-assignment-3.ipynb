## Q2 Write a python program for KNN classifier

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report,confusion_matrix
```

```python
#load data set
iris= pd.read_csv("/kaggle/input/datasets/hritamdhara/iris-iris/Iris.csv")
```

```python
#print Data shape
print(f"Data Shape: {iris.shape}")
```

```python
#Print first 5 rowa
print(f"First 5 roes: \n{iris.head()}")
```

```python
#select Features (X) and Target(Y)
x = iris.iloc[:,1:5] 
y = iris["Species"]
```

```python
#split the datset into training and testing sets
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
#create KNN model
knn=KNeighborsClassifier(n_neighbors=3)
```

```python
#Train the model
knn.fit(x_train,y_train)
```

```python
#predict on test data
y_pred=knn.predict(x_test)
```

```python
# Display results
print("\nPredicted Species:")
print(y_pred)
print("\nActual Species:")
print(y_test.values)
```

```python
#calculate accuracy
accuracy=accuracy_score(y_test,y_pred)
print("\nAccuracy:",accuracy*100,"%")
```

```python
#confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test,y_pred))
```

```python
#classfication Report
print("\nClassification Report:")
print(classification_report(y_test,y_pred))
```
